```python
import pandas as pd
import numpy as np
import os
import glob
```

```python
files = glob.glob("CPCB/*.csv")

print("Number of files:", len(files))
```

```python
dfs = []

for file in files:

    df = pd.read_csv(file)

    station = os.path.basename(file)
    station = station.replace(".csv","")

    df['Station'] = station

    dfs.append(df)

cpcb = pd.concat(dfs, ignore_index=True)

print(cpcb.shape)
```

```python
cpcb['Timestamp'] = pd.to_datetime(cpcb['Timestamp'])

cpcb['Date'] = cpcb['Timestamp'].dt.date
cpcb['Month'] = cpcb['Timestamp'].dt.month
cpcb['Day'] = cpcb['Timestamp'].dt.day
```
